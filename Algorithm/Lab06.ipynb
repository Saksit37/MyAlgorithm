## 1. Instance Simplification & Presorting

### 1. Element Uniqueness Check

```python
def is_unique(arr):
    arr = sorted(arr)

    for i in range(len(arr) - 1):
        if arr[i] == arr[i+1]:
            return False
    return True

print(is_unique([5, 5, 1, 2, 3]))
print(is_unique([10, 4, 1, 7, 9, 2]))
print(is_unique([1, 2, 3, 4, 5, 5]))
print(is_unique([100]))
```

```python
# ประสิทธิภาพด้านเวลา (Average Case)
# การ Sort ข้อมูลที่ใช้เวลา Θ(n log n)
# การตรวจสอบค่าที่ติดกันใช้เวลา Θ(n)
# ดังนั้น
# T_avg(n) = Θ(n log n) + Θ(n)
#          = Θ(n log n)

# Average Time Complexity = Θ(n log n)
# Basic Operation คือ arr[i] == arr[i + 1]
# และมีการเรียกใช้ n-1 ครั้ง
```

### 2. Find Mode using Presorting

```python
def find_mode_preson(arr):
    arr = sorted(arr)

    mode = arr[0]
    max_count = 1

    current = arr[0]
    current_count = 1

    for i in range(1, len(arr)):
        if arr[i] == current:
            current_count += 1
        else:
            if current_count > max_count:
                max_count = current_count
                mode = current

            current = arr[i]
            current_count = 1

    if current_count > max_count:
        mode = current

    return mode

print(find_mode_preson([7, 7, 7, 7]))
print(find_mode_preson([6, 5, 7, 5, 7, 5]))
print(find_mode_preson([9, 1, 4, 2]))
print(find_mode_preson([42]))
```

```python
# ประสิทธิภาพด้านเวลา (Average Case)
# Sort ใช้เวลา Θ(n log n)
# Scan ใช้เวลา Θ(n)
# ดังนั้น
# Average Time Complexity = Θ(n log n)
# Basic Operation คือ arr[i] == current
# และถูกเรียกใช้ basic operation : n - 1 ครั้ง 
```

### 3.Presorted Search Optimization

```python
def presort_search(arr, target):
    left = 0
    right = len(arr) - 1

    while left <= right:
        mid = (left + right)//2
        
        if arr[mid] == target:
            return mid
        
        elif arr[mid] < target:
            left = mid + 1
        
        else:
            right = mid - 1
    return -1

print(presort_search([5, 3, 1, 4, 2],3))    
print(presort_search([20, 10, 40, 30],40))
print(presort_search([1, 8, 3, 9],5))
print(presort_search([],10))
```

```python
# ประสิทธิภาพด้านเวลา (Average Case)
# Sort ใช้เวลา Θ(n log n)
# Binary Search ใช้เวลา Θ(log n)
# ดังนั้น
# T_avg(n) = Θ(n log n)
# Average Time Complexity = Θ(n log n)
# Basic Operation คือ arr[mid] == target
# และถูกเรียกใช้ basic operation : log n ครั้ง เพราะ binary search เป็นการลดข้อมูลครึ่งหนึ่งในแต่ละรอบ
```

### 4. Minimum Absolute Difference Pairs

```python
def min_abs_diff(arr):
    if len(arr) < 2:
        return 0

    arr = sorted(arr)
    min_diff = abs(arr[1] - arr[0])

    for i in range(1, len(arr) - 1):
        diff = arr[i + 1] - arr[i]
        if diff < min_diff:
            min_diff = diff

    return min_diff

print(min_abs_diff([4, 9, 1, 4, 15]))
print(min_abs_diff([12, 3, 22, 8, 19]))
print(min_abs_diff([100, 80, 50, 10]))
print(min_abs_diff([10, 25]))
```

```python
# ประสิทธิภาพด้านเวลา (Average Case)
# Sort ใช้เวลา Θ(n log n)
# Scan ใช้เวลา Θ(n)
# ดังนั้น
# T_avg(n) = Θ(n log n)
# Average Time Complexity = Θ(n log n)
# Basic Operation คือ การเปรียบเทียบ diff < min_diff
# และถูกเรียกใช้ basic operation : n - 1 ครั้ง
```

## 2. Gaussian Elimination & Problem Solving

### 5. Forward Elimination Phase Implementation

```python
def forward_elimination(aug_matrix):
    n = len(aug_matrix)

    for k in range(n - 1):
        for i in range(k + 1, n):
            factor = aug_matrix[i][k] / aug_matrix[k][k]
            for j in range(k, n + 1):
                aug_matrix[i][j] -= factor * aug_matrix[k][j]

    return aug_matrix

A = [[2, -1, 1],
     [4, 1, -1],
     [1, 1, 1]]

b = [1, 5, 0]

print(forward_elimination(A, b))
```

```python
# ประสิทธิภาพด้านเวลา (Average Case)
# T_avg(n) = Θ(n^3)
# ดังนั้น
# Average Time Complexity = Θ(n^3)
# Basic Operation คือ aug_matrix[i][j] -= factor * aug_matrix[k][j]
# และถูกเรียกใช้ basic operation : n^3 / 3 ครั้ง
```

## 6. Backward Substitution Solver

```python
def backward_substitution(aug_matrix):
    n = len(aug_matrix)
    x = [0] * n

    for i in range(n - 1, -1, -1):
        total = aug_matrix[i][n]
        for j in range(i + 1, n):
            total -= aug_matrix[i][j] * x[j]

        if aug_matrix[i][i] == 0:
            raise ValueError("Zero on diagonal")

        x[i] = total / aug_matrix[i][i]

    return x

aug_matrix = [
    [2, 1, 1, 1],
    [0, 3, 1, -1],
    [0, 0, 1, -1]
]

print(backward_substitution(aug_matrix))
```

```python
# ประสิทธิภาพด้านเวลา (Average Case)
# T_avg(n) = Θ(n^2)
# ดังนั้น
# Average Time Complexity = Θ(n^2)
# Basic Operation คือ aug_matrix[i][j] * x[j]
# และถูกเรียกใช้ basic operation : (n-1) + (n-2) + ... + 1 ครั้ง
```

## 3. Representation Change — Balanced Search Trees & Heaps

### 7. AVL Tree Balance Factor & Rotation Type Detector

```python
class Node:
    def __init__(self, val):
        self.val = val
        self.left = None
        self.right = None

def height(node):
        if node is None:
            return 0
        return 1 + max(height(node.left), height(node.right))


def balance_factor(node):
        if node is None:
            return 0
        return height(node.left) - height(node.right)

def check_rotation_type(root):
    if root is None:
            return "Balanced"

    bf = balance_factor(root)

    if bf == 2:
        left_bf = balance_factor(root.left)
            
        if left_bf >= 0:
            return "R-rotation"
        else:
            return "LR-rotation"
    elif bf == -2:
        right_bf = balance_factor(root.right)

        if right_bf <= 0:
            return "L-rotation"
        else:
            return "RL-rotation"

    return "Balanced"

#Single Right
root = Node(30)
root.left = Node(20)
root.left.left = Node(10)
print(check_rotation_type(root))

#Single right
root = Node(10)
root.right = Node(20)
root.right.right = Node(30)
print(check_rotation_type(root))

#Double LR
root = Node(30)
root.left = Node(10)
root.left.right = Node(20)
print(check_rotation_type(root))

#Double RL
root = Node(10)
root.right = Node(30)
root.right.left = Node(20)
print(check_rotation_type(root))

#Balanced
root = Node(30)
root.left = Node(20)
root.right = Node(40)
print(check_rotation_type(root))
```

```python
# ประสิทธิภาพด้านเวลา (Average Case)
# เนื่องจากเป็น Recursive Algorithm:
# T(n) = 2T(n/2) + Θ(1)

# ดังนั้น
# a = 2
# b = 2
# d = 0

# เพราะ
# 2^0 = 1
# และ
# a > b^d

# จาก Master Theorem:
# T(n) = Θ(n)

# ดังนั้น
# Average Time Complexity = Θ(n)
# Basic Operation คือ การคำนวณ Height/Balance Factor และการเปรียบเทียบค่า BF
# และถูกเรียกใช้ basic operation : Θ(n) ครั้ง
```

### 8.Bottom-Up Heap Construction (Heapify)

```python
def heapify_down(A, i, n):
    while True:
        left = 2 * i
        right = 2 * i + 1
        largest = i

        if left <= n and A[left] > A[largest]:
            largest = left

        if right <= n and A[right] > A[largest]:
            largest = right

        if largest == i:
            break

        A[i], A[largest] = A[largest], A[i]
        i = largest


def heap_bottom_up(A):
    n = len(A) - 1
    for i in range(n // 2, 0, -1):
        heapify_down(A, i, n)
    return A

print(heap_bottom_up([0, 10, 8, 7 ,5, 2]))
print(heap_bottom_up([0, 2, 9, 7, 6, 5, 8]))
print(heap_bottom_up([0, 1, 2, 3, 4, 5, 6]))
print(heap_bottom_up([0, 5]))
```

```python
# ประสิทธิภาพด้านเวลา (Average Case)
# แม้ heapify_down บาง Node จะทำงานหลายระดับ แต่ Bottom-Up Heap Construction มีผลรวม: Θ(n)
# T_avg(n) = Θ(n)

# Average Time Complexity = Θ(n)
# Basic Operation คือ 
# การเปรียบเทียบ: A[left] > A[largest] และ A[right] > A[largest]
# และถูกเรียกใช้ basic operation : ประมาณ Θ(n) ครั้ง
```

### 9. Heap Root Deletion (Pop Max)

```python
def heap_delete_root(H):
    n = len(H) - 1
    if n == 0:
        return None, H
    max_val = H[1]

    if n == 1:
        return max_val, [0]
    H[1] = H[n]
    H.pop()

    heapify_down(H, 1, n - 1)

    return max_val, H

H = [0, 9, 6, 8, 2, 5, 7]

max_val, H = heap_delete_root(H)

print("คืน", max_val)
print("Heap ใหม่ คือ", H)
```

```python
# ประสิทธิภาพด้านเวลา (Average Case)
# ความสูงของ Heap คือ Θ(log n)
# ดังนั้น Shift Down ใช้ Θ(log n)
# จะได้
# T_avg(n) = Θ(log n)
# Average Time Complexity = Θ(log n)
# Basic Operation คือ การเปรียบเทียบ Parent กับ Child ระหว่าง Shift Down
# และถูกเรียกใช้ basic operation : ประมาณ 2 log n ครั้ง
```

### 10. Heap Sort Complete Implementation

```python
def heapsort(A):
    h = [0] + A[:]

    heap_bottom_up(h)

    result = []

    while len(h) > 1:
        max_val, h = heap_delete_root(h)
        result.append(max_val)

    result.reverse()

    return result

print(heapsort([2, 9, 7, 6, 5, 8]))
print(heapsort([1, 2, 3, 4, 5]))
print(heapsort([5, 4, 3, 2, 1]))
print(heapsort([4, 2, 4, 1, 2]))
```

```python
# ประสิทธิภาพด้านเวลา (Average Case)
# สร้าง Heap: Θ(n)
# ลบ Root n-1 ครั้ง: Θ(n log n)
# ดังนั้น
# T_avg(n) = Θ(n log n)
# Average Time Complexity = Θ(n log n)
# Basic Operation คือ การเปรียบเทียบ Parent/Child ระหว่าง Heapify
# และถูกเรียกใช้ basic operation : Θ(n log n)
```

## 4. Representation Change — Horner's Rule & Binary Exponentiation

### 11. Polynomial Evaluation via Horner's Rule 

```python
def horner(P, x):
    result = P[-1]

    for i in range(len(P) - 2, -1, -1):
        result = result * x + P[i]

    return result

print(horner([-5, 1, 3, -1, 2], 3))
print(horner([3, 2], 5))
print(horner([7], 100))
```

```python
# ประสิทธิภาพด้านเวลา (Average Case)
# จำนวนรอบไม่ขึ้นกับค่าของข้อมูล : T_avg(n) = Θ(n)
# Average Time Complexity = Θ(n)
# Basic Operation คือ result * x + P[i]
# และถูกเรียกใช้ basic operation : n ครั้ง
```

### 12. Left-to-Right Binary Exponentiation

```python
def left_right_binary_exp(a, b_binary):
    result = 1

    for bit in b_binary:
        result = result * result
        if bit == 1:
            result = result * a

    return result

print(left_right_binary_exp(3, [1, 1, 0, 1]))
print(left_right_binary_exp(2, [1, 0, 0, 0]))
print(left_right_binary_exp(5, [1]))
print(left_right_binary_exp(9, [0]))
```

```python
# ประสิทธิภาพด้านเวลา (Average Case)
# โดยเฉลี่ยประมาณครึ่งหนึ่งของ Bits เป็น 1
# ดังนั้นจำนวนการคูณโดยประมาณ: k squaring operations + k/2 multiplications ≈ 1.5k
# T_avg(n) = Θ(log n)

# Average Time Complexity = Θ(log n)
# Basic Operation คือ result * result และเมื่อ Bit = 1: result * a
# และถูกเรียกใช้ basic operation : log_2 n ครั้ง
```

### 13. Right-to-Left Binary Exponentiation

```python
def right_left_binary_exp(a, b_binary):
    term = a
    product = 1

    for bit in b_binary:
        if bit == 1:
            product *= term

        term *= term

    return product

print(right_left_binary_exp(3, [1, 0, 1, 1]))
print(right_left_binary_exp(10, [1]))
print(right_left_binary_exp(4, [0, 1, 0]))
print(right_left_binary_exp(2, [1, 1, 1]))
```

```python
# ประสิทธิภาพด้านเวลา (Average Case)
# ถ้ามี k Bits: k = Θ(log n)
# โดยเฉลี่ยประมาณ k/2 Bits เป็น 1
# ดังนั้นยังเป็น: T_avg(n) = Θ(log n)

# Average Time Complexity = Θ(log n)
# Basic Operation คือ product *= term และ term *= term
# และถูกเรียกใช้ basic operation : log_2 n ครั้ง
```

## 5. Problem Reduction

### 14. Least Common Multiple (LCM) via GCD Reduction

```python
def gcd(m, n):
    while n != 0:
        m, n = n, m % n
    return abs(m)


def lcm(m, n):
    if m == 0 or n == 0:
        return 0
    return abs(m * n) // gcd(m, n)

print(lcm(24, 60))
print(lcm(17, 13))
print(lcm(5, 25))
print(lcm(42, 42))
```

```python
# ประสิทธิภาพด้านเวลา (Average Case)
# Euclid's Algorithm มีจำนวนรอบเติบโตประมาณ: Θ(log(min(m, n)))
# T_avg(n) = Θ(log(min(m,n)))

# Average Time Complexity = Θ(log(min(m,n)))
# Basic Operation คือ m % n
# และถูกเรียกใช้ basic operation : n ครั้ง
```

### 15. State-Space Graph Reduction

```python
from collections import deque

def solve_river_puzzle():
    start = (0, 0, 0, 0)
    goal = (1, 1, 1, 1)

    def is_safe(state):
        farmer, wolf, goat, cabbage = state

        # Wolf อยู่กับ Goat โดยไม่มี Farmer
        if wolf == goat and farmer != wolf:
            return False

        # Goat อยู่กับ Cabbage โดยไม่มี Farmer
        if goat == cabbage and farmer != goat:
            return False

        return True

    def get_next_states(state):
        farmer, wolf, goat, cabbage = state
        states = []

        # Farmer ข้ามคนเดียว
        next_farmer = 1 - farmer
        new_state = (next_farmer, wolf, goat, cabbage)

        if is_safe(new_state):
            states.append(new_state)

        # Farmer พา Wolf
        if wolf == farmer:
            new_state = (
                next_farmer,
                next_farmer,
                goat,
                cabbage
            )

            if is_safe(new_state):
                states.append(new_state)

        # Farmer พา Goat
        if goat == farmer:
            new_state = (
                next_farmer,
                wolf,
                next_farmer,
                cabbage
            )

            if is_safe(new_state):
                states.append(new_state)

        # Farmer พา Cabbage
        if cabbage == farmer:
            new_state = (
                next_farmer,
                wolf,
                goat,
                next_farmer
            )

            if is_safe(new_state):
                states.append(new_state)

        return states

    queue = deque([(start, [start])])
    visited = {start}

    while queue:
        state, path = queue.popleft()

        if state == goal:
            return path

        for next_state in get_next_states(state):
            if next_state not in visited:
                visited.add(next_state)
                queue.append((next_state, path + [next_state]))

    return []

path = solve_river_puzzle()

for state in path:
    print(state)
```

```python
# ประสิทธิภาพด้านเวลา (Average Case)
# T_avg(n) = O(V + E) โดย V ≤ 16

# Average Time Complexity = O(V + E)
# Basic Operation คือ การตรวจสอบว่า State ปลอดภัยหรือไม่ เช่น
#                                                   wolf == goat and farmer != wolf หรือ goat == cabbage and farmer != goat
# และถูกเรียกใช้ basic operation : BFS ตรวจสอบ States และ Edges จึงมีจำนวนงานในระดับ O(V + E)
```
