## ข้อ1.1 TSP

```python
from itertools import permutations  #itertools เป็น Module ทีมีเครื่องมือช่วยจัดการข้อมูลการวนซ้ำ || itertools = กล่องเครื่องมือที่ Python เตรียมไว้ให้ สำหรับสร้าง/จัดการลำดับข้อมูลต่าง ๆ
# permutations การสับเปลี่ยนโดยสนใจลำดับ
# ระยะทางระหว่างจุด
distance = {
    'A': {'B': 4, 'C': 7, 'D': 9, 'E': 5},
    'B': {'A': 4, 'C': 3, 'D': 6, 'E': 8},
    'C': {'A': 7, 'B': 3, 'D': 2, 'E': 6},
    'D': {'A': 9, 'B': 6, 'C': 2, 'E': 4},
    'E': {'A': 5, 'B': 8, 'C': 6, 'D': 4}
}

def tsp_exhaustive(points):
    best_route = None
    best_distance = float('inf')

    # ลองทุกลำดับของจุด
    for route in permutations(points):
        total_distance = 0

        # A → จุดแรก
        total_distance += distance['A'][route[0]]

        # เดินจากจุดหนึ่งไปอีกจุด
        for i in range(len(route) - 1):
            total_distance += distance[route[i]][route[i + 1]]

        # จุดสุดท้าย → A
        total_distance += distance[route[-1]]['A']

        # เก็บเส้นทางที่สั้นที่สุด
        if total_distance < best_distance:
            best_distance = total_distance
            best_route = route

    return best_route, best_distance


points = ['B', 'C', 'D', 'E']

route, total = tsp_exhaustive(points)

print("เส้นทางที่สั้นที่สุด:")
print("A →", " → ".join(route), "→ A")
print("ระยะทางรวม:", total)
```

## ข้อ1.2 Knapsack 

```python
from itertools import combinations
# permutations การจัดหมู่ แบบไใาว
packages = [
    ("P1", 30, 400),
    ("P2", 25, 700),
    ("P3", 20, 900),
    ("P4", 35, 500),
    ("P5", 15, 400)
]

capacity = 80
best_packages = None
best_weight = 0
best_value = 0

# ลองทุกชุดของพัสดุ
for r in range(len(packages) + 1):
    for selected in combinations(packages, r):
        total_weight = sum(item[1] for item in selected)
        total_value = sum(item[2] for item in selected)

        # ต้องไม่เกิน 80 kg
        if total_weight <= capacity:
            if total_value > best_value:
                best_packages = selected
                best_weight = total_weight
                best_value = total_value


print("พัสดุที่เลือก:")

for package in best_packages:
    print(package[0])

print("น้ำหนักรวม:", best_weight, "kg")
print("ค่าส่งรวม:", best_value, "บาท")
```

# ข้อ 2 จัดตารางเวรพยาบาล (Assignment Problem)

```python
from itertools import permutations

# คะแนนของพยาบาลแต่ละคนในแต่ละกะ
score = [
    [3, 5, 7, 2],  # นิด
    [6, 2, 4, 8],  # หน่อย
    [4, 7, 3, 5],  # นุ่น
    [2, 6, 5, 4]   # เนย
]

nurses = ["นิด", "หน่อย", "นุ่น", "เนย"]
shifts = ["เช้า", "บ่าย", "ดึก", "พัก"]

best_score = float("inf")
best_assignment = None

# ลองจัดกะทุกแบบ
for assignment in permutations(range(4)):

    total = 0

    for i in range(4):
        total += score[i][assignment[i]]

    # คะแนนน้อยที่สุด = ดีที่สุด
    if total < best_score:
        best_score = total
        best_assignment = assignment


# แสดงผล
print("ผลการจัดกะที่ดีที่สุด")

for i in range(4):
    print(nurses[i], "->", shifts[best_assignment[i]])

print("คะแนนรวม =", best_score)
```

# ข้อ 3 — วางต าแหน่งสถานีชาร์จรถไฟฟ้า (Closest-Pair + Convex Hull)

```python
from math import sqrt

points = {
    "P1": (2, 3),
    "P2": (5, 8),
    "P3": (9, 2),
    "P4": (1, 7),
    "P5": (8, 9),
    "P6": (6, 4)
}

names = list(points)

best_distance = float("inf")
best_pair = None

for i in range(len(names)):
    for j in range(i + 1, len(names)):
        x1, y1 = points[names[i]]
        x2, y2 = points[names[j]]

        d = sqrt((x1 - x2) ** 2 + (y1 - y2) ** 2)

        if d < best_distance:
            best_distance = d
            best_pair = (names[i], names[j])

print("คู่ที่ใกล้ที่สุด =", best_pair)
print("ระยะห่าง =", best_distance)
```

# ข้อ 4 วางแผนลงทะเบียนเรียน (Topological Sort + Insertion Sort)

```python
from itertools import permutations

nodes = ["V1", "V2", "V3", "V4", "V5", "V6", "V7"]

edges = [
    ("V1", "V2"),
    ("V1", "V3"),
    ("V2", "V4"),
    ("V3", "V4"),
    ("V3", "V5"),
    ("V4", "V6"),
    ("V5", "V6"),
    ("V6", "V7")
]

valid_orders = []

for order in permutations(nodes):
    position = {node: i for i, node in enumerate(order)}

    if all(position[a] < position[b] for a, b in edges):
        valid_orders.append(order)

for order in valid_orders:
    print(order)
```

# ข้อ 6 เกม Nim ในงานเลีย้งบริษัท (Variable-Size Decrease)

```python
def nim_sum(piles):
    result = 0

    for pile in piles:
        result = result ^ pile

    return result


piles = [5, 3, 7]

result = nim_sum(piles)

print("Nim-sum =", result)

if result != 0:
    print("ผู้เล่นคนแรกชนะ")
else:
    print("ผู้เล่นคนที่สองชนะ")
```

เพราะ Nim-sum ≠ 0 ผู้เล่นแรกมี winning strategy

6.2 ผู้เล่นแรกหยิบ C ออก 4: (5,3,7) → (5,3,3)

Nim-sum: 5 XOR 3 XOR 3 = 5

ผู้เล่นที่สองต้องทำให้ Nim-sum = 0

ลด A จาก 5 → 0: (0,3,3)

0 XOR 3 XOR 3 = 0

ดังนั้นผู้เล่นที่สองควรหยิบ A ออก 5 ก้อน

6.3 Variable-Size-Decrease คือในแต่ละรอบขนาดปัญหาลดลงไม่เท่ากัน เช่น อาจหยิบ 1, 2, 3 หรือหลายก้อน

# ข้อ 7 จัดเรียงธุรกรรมรายวัน (Mergesort vs Quicksort)

```python
# Merge Sort
def merge_sort(arr):
    if len(arr) <= 1:
        return arr

    mid = len(arr) // 2

    left = merge_sort(arr[:mid])
    right = merge_sort(arr[mid:])

    result = []
    i = 0
    j = 0

    while i < len(left) and j < len(right):
        if left[i] <= right[j]:
            result.append(left[i])
            i += 1
        else:
            result.append(right[j])
            j += 1

    result.extend(left[i:])
    result.extend(right[j:])

    return result


data = [950, 230, 780, 120, 560, 340, 890, 450, 670]
print(merge_sort(data))
```

```python
# Quick Sort
def quick_sort(arr):
    if len(arr) <= 1:
        return arr

    pivot = arr[len(arr) // 2]

    left = [x for x in arr if x < pivot]
    middle = [x for x in arr if x == pivot]
    right = [x for x in arr if x > pivot]

    return quick_sort(left) + middle + quick_sort(right)


data = [950, 230, 780, 120, 560, 340, 890, 450, 670]
print(quick_sort(data))
```

# ข้อ 8 โครงสร้างองค์กรบริษัทขายตรง (Binary Tree Traversal)

```python
class Node:
    def __init__(self, value, sales):
        self.value = value
        self.sales = sales
        self.left = None
        self.right = None

def preorder(root):
    if root is None:
        return

    print(root.value, end=" ")
    preorder(root.left)
    preorder(root.right)

def inorder(root):
    if root is None:
        return

    inorder(root.left)
    print(root.value, end=" ")
    inorder(root.right)

def postorder_sum(root):
    if root is None:
        return 0

    left_sum = postorder_sum(root.left)
    right_sum = postorder_sum(root.right)

    return left_sum + right_sum + root.sales

root = Node("ก", 5)

root.left = Node("ข", 8)
root.right = Node("ค", 6)

root.left.left = Node("ง", 12)
root.left.right = Node("จ", 9)

root.right.left = Node("ฉ", 7)
root.right.right = Node("ช", 15)


print("Preorder:", end=" ")
preorder(root)
print()

print("Inorder:", end=" ")
inorder(root)
print()

print("Postorder:", end=" ")
def postorder(root):
    if root is None:
        return

    postorder(root.left)
    postorder(root.right)
    print(root.value, end=" ")

postorder(root)
print()

print("ยอดขายรวม =", postorder_sum(root), "ล้านบาท")

n = 7
A = n
C = 2 * n + 1
external_nodes = n + 1

print("จำนวนการบวก A(n) =", A)
print("จำนวนการเปรียบเทียบ C(n) =", C)
print("จำนวน External Nodes =", external_nodes)
```

# ข้อ 9 หาสาขาธนาคารที่ใกล้กันที่สุด (Closest-Pair by Divide-and-Conquer)

```python
from math import sqrt

points = [
    (1, 2), (3, 7), (5, 1), (8, 6),
    (2, 9), (7, 3), (9, 8), (4, 5)
]

best = float("inf")
best_pair = None

for i in range(len(points)):
    for j in range(i + 1, len(points)):
        d = sqrt(
            (points[i][0] - points[j][0]) ** 2 +
            (points[i][1] - points[j][1]) ** 2
        )

        if d < best:
            best = d
            best_pair = (points[i], points[j])

print(best_pair)
print(best)
```

# ข้อ 10 

```python
def gaussian_elimination(A, B):
    n = len(A)
    matrix = []

    for i in range(n):
        matrix.append(A[i] + [B[i]])

    for k in range(n - 1):

        for i in range(k + 1, n):
            factor = matrix[i][k] / matrix[k][k]

            for j in range(k, n + 1):
                matrix[i][j] -= factor * matrix[k][j]

    x = [0] * n

    for i in range(n - 1, -1, -1):
        total = 0

        for j in range(i + 1, n):
            total += matrix[i][j] * x[j]

        x[i] = (matrix[i][n] - total) / matrix[i][i]

    return x


A = [
    [2, -1, 1],
    [4, 1, -1],
    [1, 1, 1]
]

B = [1, 5, 0]


solution = gaussian_elimination(A, B)
print("x =", solution[0])
print("y =", solution[1])
print("z =", solution[2])
```

# ข้อ 11 จัดลำดับผู้ป่วยฉุกเฉิน (Heap Sort + AVL Tree)

```python
def heapify(arr, n, i):
    largest = i
    left = 2 * i + 1
    right = 2 * i + 2

    if left < n and arr[left] > arr[largest]:
        largest = left

    if right < n and arr[right] > arr[largest]:
        largest = right

    if largest != i:
        arr[i], arr[largest] = arr[largest], arr[i]
        heapify(arr, n, largest)


def build_max_heap(arr):
    n = len(arr)

    for i in range(n // 2 - 1, -1, -1):
        heapify(arr, n, i)


data = [45, 88, 12, 95, 67, 34, 71, 53]

build_max_heap(data)

print(data)
```

# ข้อ 12 คำนวณดอกเบี้ย ทบต้น + ลดปัญหา (Horner's Rule + Problem Reduction)

```python
def horner(P, x):
    result = P[-1]

    for i in range(len(P) - 2, -1, -1):
        result = result * x + P[i]

    return result


P = [10, 2, -7, 3, 5]

print(horner(P, 4))
```

```python
def gcd(a, b):
    while b != 0:
        a, b = b, a % b

    return a


def lcm(a, b):
    return (a * b) // gcd(a, b)


print(gcd(24, 60))
print(lcm(24, 60))
```

การ Reduction ช่วยเปลี่ยนปัญหา LCM ให้เป็นปัญหา GCD ซึ่งมี Algorithm ที่มีประสิทธิภาพ
