# Big O

## $เริ่มจากที่ง่ายที่สุด O(1) จำประโยคนี้ : O(1) ข้อมูลมีกี่ตัวก็ไม่สน ทำงานจำนวนครั้งคงที่$

### Example

```python
numbers = [10, 20, 30, 40, 50]
print(numbers[0]) 
#เราเข้าถึง numbers[0] โดยตรง ดังนั้นจึงเป็น O(1)
```

```python
x = 10
y = 10

result = x + y #ก็เป็น O(1) เหมือนกัน เพราะเป็นการบวกตัวเลขสองตัวโดยตรง
print(result)
```

## $O(n) จำประโยคนี้ : O(n) = ยิ่งข้อมูลเยอะ ยิ่งต้องทำงานตามจำนวนข้อมูล$

### Example

```python
numbers = [10, 20, 30, 40, 50]

for number in numbers:
    print(number)

#สมมติว่า n = 5 Loop ทำงานประมาณ 5 ครั้ง ถ้า n = 100 ก็ทำงานประมาณ 100 ครั้ง และถ้า n = 1000 ก็ทำงานประมาณ 1000 ครั้ง ดังนั้นเราจึงบอกว่า Loop นี้เป็น O(n)
```

## $O(n^2) ก็เริ่มจากตรงนี้$

### Example

```python
numbers = [10, 20, 30, 40, 50]

for i in numbers:
    for j in numbers:
        print(i, j)

#มี Loop ซ้อนกัน ถ้า n = 5 ก็จะได้ 5 * 5 = 25 นั้นก็คือทำงาน 25 ครั้ง ถ้า n = 100 ก็จะได้ 100 * 100 = 10,000 ครั้ง และถ้า n = 1000 ก็จะได้ 1,000 * 1,000 = 1,000,000 ครั้ง ดังนั้นเราจึงบอกว่า Loop นี้เป็น O(n^2) 
```

## $ O(log_n) คืออะไร$ อันนี้แหละที่คนเริ่ม Algorithm มักงง ลองโจทย์ดู
มีตัวเลข 1- 100 ให้หาเลขที่เราต้องการ ถ้าใช้วิธีธรรมดา ก็จะต้องค้นถึง 100 ครั้ง เพื่อหาเลขที่ต้องการ

แต่ถ้าใช้ Binary Search เราจะไม่ค้นทีละตัว เช่น 

ต้องการหาเลข 73 จะเริ่มตรงกลาง : $1 ─────── 50 ─────── 100$ ซึ่ง 73 มากกว่า 50 ฉะนั้นสามารถทิ้งฝั่งที่เป็นตัวเลข 1 - 50 ได้เลย 

ตอนนี้จะเหลือ : $51 ─────── 75 ─────── 100$ ซึ่ง 73 น้อยกว่า 75 ฉะนั้นสามารถทิ้งฝั่งที่เป็นตัวเลข 76 - 100 ได้เลย

ทุกครั้งเราจะตัดข้อมูลออกครึ่งหนึ่ง นี่คือ แนวคิดของ $O(log_n)$

```python
n = 100
while n > 1:
    n = n // 2
    print(n)

# แต่ละรอบ n ถูกหาร 2 ลงเรื่อยๆ
```

### Binary Search อันนี้ต่างออกไปมาก ข้อมูลต้องเรียงก่อน เช่น :
$[20, 12, 5, 21, 64, 1, 88] → [1, 5, 12, 20, 21, 64, 88]$ ต้องการหา 21 เราจะไม่ไล่จากซ้าย แต่เราจะดูตรงการเลย ในที่นี้ คือเลข 20

จะได้ $21>20$ ดังนั้นจะตัดด้านซ้ายได้เลย ได้แก่ $1,5,12$ ดังนั้นจะเหลือ $21,64,88$ ดูตรงกลาง จะได้ $21<64$ ดังนั้นจะตัดด้านขวาได้เลย ได้แก่ $88$ ดังนั้นก็จะเจอพอดี

แล้วทำไมถึงเป็น $ O(log n)$ เพราะทุกครั้งมันจะลดครึ่งหนึ่ง $n, n/2, n/4, n/8, n/16$ จำง่ายๆ ตัดครึ่งทุกครั้ง คือ Binary Search นี่เป็น Plattern ที่สำคัญมากของ Algorithm

## Selection Sort
จะคิดว่า "เอาตัวที่เล็กที่สุดไปไว้ข้างหน้า" เช่น $[5, 3, 8, 1, 2]$ จะได้ :

$รอบแรก [5, 3, 8, 1, 2] จะได้ ค่า min คือ 1 ฉะนั้นเอาเลข 1 ไปไว้ข้างหน้า จะได้ : [1, 5, 3, 8, 2]$

$รอบที่สอง [1, 5, 3, 8, 2] จะได้ ค่า min คือ 2 ฉะนั้นเอาเลข 2 ไปไว้ข้างหน้า จะได้ : [1, 2, 5, 3, 8]$

$รอบที่สาม [1, 2, 5, 3, 8] จะได้ ค่า min คือ 3 ฉะนั้นเอาเลข 3 ไปไว้ข้างหน้า จะได้ : [1, 2, 3, 5, 8] สิ้นสุดพอดี เพราะมันเรียงเลขกันแล้ว$

แล้วทำไมถึงเป็น $O(n^2)$ ลองดูจำนวนที่ต้องตรวจ

$รอบแรก : n$

$รอบสอง : n - 1$

$รอบสาม : n - 2$

ดังนั้นจะได้เป็น $n + (n+1) + (n+2) + ... + (n+m)$ สุดท้ายมันโตประมาณ : $n^2$ ดังนั้น $Selection \;Sort = O(n^2)$

## จำง่ายๆ หา Max/Min แล้วไล่หาทุกครั้ง มักจะเป็น $O(n^2)$

## Insertion Sort อันนี้อยากให้จำเป็นการเรียงไพ่ในมือ มากกว่า
สมมติไพ่(รอบแรก) : 5 | 3 8 1 2 คือบนมือเราเป็นเลข 5 หยิบเลข 3 มา จะเรียงได้ 3 5 | 8 1 2

รอบที่สอง : 3 5 | 8 1 2 คือบนมือเรามีเลข 3 5 หยิบ 8 มา จะเรียงได้ 3 5 8 | 1 2

รอบที่สาม : 3 5 8 | 1 2 คือบนมือเรามีเลข 3 5 8 หยิบ 1 มา จะเรียงได้ 1 3 5 8 | 2

รอบที่สี่ : 1 3 5 8 | 2 คือบนมือเรามีเลข 1 3 5 8 หยิบ 2 มา จะเรียงได้ 1 2 3 5 8 |

$กรณีนี้เป็น Worst case เพราะเราอาจจะต้องเลื่อนหรือเปรียบเทียบหลายตัว เช่นข้อมูล เรียงย้อนกลับ 5 4 3 2 1 มันต้องขยับเยอะมากจึงได้เป็น \;Worst case: Insertion\;Sort = O(n^2)$

## แต่ถ้า มีข้อมูล เกือบเรียงกันครบอยู่แล้ว Insertion Sort นี้จะได้ทำงานได้ดีมาก เป็น Best Case
## $ให้จำว่า Best case = O(n),\;Worst case = O(n^2)$

## Merge Sort ตัวนี้ก็สำคัญมาก เพราะเป็นตัวแรกที่จะเห็นว่า แบ่งปัญหาออกเป็นครึ่งๆ
เช่น สมมติ : $[8, 3, 5, 4, 7, 6, 1, 2]$ จะแบ่งได้เป็น

$[8, 3, 5, 4]\quad[7, 6, 1, 2]$ แบ่งอีกก็จะได้เป็น: $[8, 3]\;[5, 4]\quad[7, 6]\;[1, 2]$ แบ่งอีกก็จะได้: $[8]\;[3]\;[5]\;[4]\;[7]\;[6]\;[1]\;[2]$

จากนั้นค่อยๆ Merge กลับ จะได้ $[3, 4, 5, 8]\quad[1, 2, 6, 7]$ และ $[3, 8]\;[4, 5]\quad[6, 7]\;[1, 2]$ และ $[1]\;[2]\;[3]\;[4]\;[5]\;[6]\;[7]\;[8]$

### นี่คือ Divide and Conquer

แล้วทำไม Merge Sort ถึงเป็น $O(n\; log n)$
เพราะมันมี 2 ส่วน:

ส่วนที่ 1 : เป็นการแบ่งครึ่ง คือ log n

ส่วนที่ 2 : เป็นการเข้าถึงข้อมูลโดยรวมโดยตรง คือ n

ดังนั้น Merge Sort = $O(n\; log_n)$

## Bubble Sort
แนวคิดหลัก คือ เปรียบเทียบตัวที่อยู่ติดกัน แล้วสลับถ้าลำดับผิด เช่น: 

จาก [5, 3, 8, 1, 2] ให้เรียงจากน้อยไปมาก ก่อน

คู่แรก : 5 > 3 ให้สลับ จะได้ [3, 5, 8, 1, 2] 

ต่อไปคู่ที่สอง : 5 < 8 ถูกต้องจึงไม่ต้องสลับ

ต่อไปคู่ที่สาม : 8 > 1 ให้สลับ จะได้ [3, 5, 1, 8, 2]

ต่อไปคู่ที่สี่ : 8 > 2 ให้สลับ จะได้ [3, 5, 1, 2, 8] สังเกตมั้ยว่า 8 ซึ่งเป็นค่าที่มากที่สุดถูก "ดัน" ไปอยู่เลขท้ายสุด

## นี่คือที่มาของ Bubble Sort ค่าที่ใหญ่ๆจะค่อยๆ ถูกดันไปอยู่ด้านท้าย
## เป็น $O(n^2)$ - Worst case, Average case
## Best case เป็น $O(n)$

## Quick Sort ใช้แนวคิด Divide and Conquer
หลักๆ จะมี 3 ขั้น : 
1. เลือก Pivot

2. Partition

3. ทำซ้ำกับซ้ายและขวา

### 1. Pivot
$สมมติ : [8, 3, 5, 4, 7, 6, 1, 2]\; เลือกเลข 5 จากนั้นจะแบ่งข้อมูลเป็น ค่าที่น้อยกว่า\; 5 \;และค่าที่มากกว่า\; 5 --> [?\, < \,5 \,|\, 5 \,> \,?]$

$จะได้\; [3, 4, 1, 2] \;|\; 5 \;|\; [8, 7, 6]\; นี่คือ หัวมจของ Quick Sort$

### 2. Partition
เราจะเอาฝั่งซ้าย และขวา ไป Quick Sort ต่อ

```python
## จะได้แบบนี้
# [8, 3, 5, 4, 7, 6, 1, 2]

#                   │
#                pivot=5
#              /         \
#       [3,4,1,2]       [8,7,6]
#          │                │
#        pivot              pivot
#        /   \              /   \
#       ...  ...           ...  ...

## นี่คือ Divide and Conquer
```

```python
## Linear Search ตรวจเลขใน list กับ target

def linear_search(arr, target):

    for i in range(len(arr)):
        if arr[i] == target:
            return i
    return -1

#บอกตำแหน่งว่าตำแหน่งไหน ตรงกับ Target
arr = [30, 25, 10, 50, 45]
print(linear_search(arr, 45))
```

```python
## Binary Search
# หาตัวเลขที่ต้องการ target ว่าอยู่ใน index ที่เท่าไหร่

# def binary_search(arr, target):

#     left = 0
#     right = len(arr) - 1

#     while left <= right:
#         mid = (left + right) // 2
#         if arr[mid] == target:
#             return mid

#         elif arr[mid] < target:
#             left = mid + 1

#         else:
#             right = mid - 1
#     return -1

# arr = [10 ,20 ,30, 40, 50, 60, 70]
# print(binary_search(arr, 60))

def binary_search(arr, target):
    left = 0
    right = len(arr) - 1

    while left <= right:
        mid = (left + right)//2

        if arr[mid] == target:
            return mid

        elif arr[mid] < target:
            left = mid + 1

        else:
            right = mid - 1
    return -1

arr = [3, 7, 12, 18, 25, 31, 40, 55, 70]
print(binary_search(arr, 31))
```
