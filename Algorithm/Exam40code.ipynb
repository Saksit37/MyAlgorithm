### 1. Two Sum Problem (การหาคู่บวก)
Time: O(N) — วนลูปครั้งเดียว, การค้นหาใน dict เฉลี่ย O(1)

Space: O(N) — เก็บค่าที่เคยเจอใน hash table

```python
def two_sum(nums, target):
    seen = {}  # value -> index
    for i, num in enumerate(nums):
        complement = target - num
        if complement in seen:
            return [seen[complement], i]
        seen[num] = i
    return [] 

print(two_sum([2, 7, 11, 15], 9))  # [0, 1]
```

### 2. Fibonacci Dynamic Programming (Memoization)
Time: O(N) — แต่ละค่าคำนวณครั้งเดียวแล้วจำไว้

Space: O(N) — เก็บ memo dict + recursion stack

```python
def fib_memo(n, memo=None):
    if memo is None:
        memo = {}
    if n <= 1:
        return n
    if n in memo:
        return memo[n]
    memo[n] = fib_memo(n - 1, memo) + fib_memo(n - 2, memo)
    return memo[n]

print(fib_memo(30))  # 832040
```

### 3. Counting Sort implementation
Time: O(N + K) — N คือขนาด array, K คือช่วงข้อมูล (0-1000)

Space: O(K) — frequency array ขนาด K+1

```python
def counting_sort(arr, k=1000):
    count = [0] * (k + 1)
    for num in arr:
        count[num] += 1

    result = []
    for value, freq in enumerate(count):
        result.extend([value] * freq)
    return result

print(counting_sort([4, 2, 2, 8, 3, 3, 1]))
```

### 4. First Non-Repeating Character (ตัวอักษรแรกที่ไม่ซ้ำ)
Time: O(N) — วน 2 รอบ (ไม่ซ้อนกัน) = O(2N) = O(N)

Space: O(1) หรือ O(K) — K คือจำนวนตัวอักษรที่เป็นไปได้ (คงที่)

```python
def first_non_repeating(s):
    freq = {}
    for ch in s:               # รอบที่ 1: นับความถี่
        freq[ch] = freq.get(ch, 0) + 1

    for ch in s:                # รอบที่ 2: หาตัวแรกที่ freq = 1
        if freq[ch] == 1:
            return ch
    return None

print(first_non_repeating("swiss"))  # 'w'
```

### 5. Range Sum Query (Prefix Sum Array)
Time: O(N) สร้างครั้งเดียว, O(1) ต่อ query

Space: O(N) — เก็บ prefix sum array

```python
class RangeSumQuery:
    def __init__(self, arr):
        self.prefix = [0] * (len(arr) + 1)
        for i in range(len(arr)):
            self.prefix[i + 1] = self.prefix[i] + arr[i]

    def query(self, l, r):  # inclusive [l, r]
        return self.prefix[r + 1] - self.prefix[l]

rsq = RangeSumQuery([2, 4, 5, 1, 6, 3])
print(rsq.query(1, 3))  # 4+5+1 = 10
```

### 6. Anagram Detection
Time: O(N) — วน string ทั้งสองครั้งละหนึ่งรอบ

Space: O(K) — K คือจำนวนตัวอักษรที่ไม่ซ้ำ (คงที่หากเป็นตัวอักษรจำกัด)

```python
def is_anagram(s1, s2):
    if len(s1) != len(s2):
        return False

    count = {}
    for ch in s1:
        count[ch] = count.get(ch, 0) + 1
    for ch in s2:
        if ch not in count:
            return False
        count[ch] -= 1
        if count[ch] == 0:
            del count[ch]

    return len(count) == 0

print(is_anagram("listen", "silent"))  # True
```

### 7. Find Duplicate Number in Array
Time: O(N) — วนลูปครั้งเดียว

Space: O(N) — เก็บ visited set

```python
def find_duplicate(nums):
    seen = set()
    for num in nums:
        if num in seen:
            return num
        seen.add(num)
    return -1

print(find_duplicate([1, 3, 4, 2, 2]))  # 2
```

### 8. Longest Consecutive Sequence
Time: O(N) — แต่ละตัวเลขถูกเข้าไปเช็คใน while loop รวมกันไม่เกิน N ครั้ง (amortized)

Space: O(N) — hash set เก็บทุกตัวเลข

```python
def longest_consecutive(nums):
    num_set = set(nums)
    longest = 0

    for num in num_set:
        if num - 1 not in num_set:  # เป็นจุดเริ่มต้นของ sequence เท่านั้น
            length = 1
            while num + length in num_set:
                length += 1
            longest = max(longest, length)

    return longest

print(longest_consecutive([100, 4, 200, 1, 3, 2]))  # 4 (1,2,3,4)
```

### 9. Matrix 2D Sum Query (2D Prefix Sum)
Time: O(MN) สร้างครั้งเดียว, O(1) ต่อ query

Space: O(MN) — 2D prefix sum matrix

```python
class Matrix2DSum:
    def __init__(self, matrix):
        m, n = len(matrix), len(matrix[0])
        self.prefix = [[0] * (n + 1) for _ in range(m + 1)]
        for i in range(m):
            for j in range(n):
                self.prefix[i+1][j+1] = (matrix[i][j]
                    + self.prefix[i][j+1]
                    + self.prefix[i+1][j]
                    - self.prefix[i][j])

    def query(self, r1, c1, r2, c2):  # sub-matrix (r1,c1) ถึง (r2,c2)
        return (self.prefix[r2+1][c2+1]
                - self.prefix[r1][c2+1]
                - self.prefix[r2+1][c1]
                + self.prefix[r1][c1])

matrix = [[1,2,3],[4,5,6],[7,8,9]]
m2d = Matrix2DSum(matrix)
print(m2d.query(0, 0, 1, 1))  # 1+2+4+5 = 12
```

### 10. Subarray Sum Equals K
Time: O(N) — วนลูปครั้งเดียว, hash map lookup O(1)

Space: O(N) — เก็บความถี่ของ prefix sum

```python
def subarray_sum(nums, k):
    count = 0
    prefix_sum = 0
    freq = {0: 1}  # prefix sum = 0 เจอมาแล้ว 1 ครั้ง (กรณี subarray เริ่มจาก index 0)

    for num in nums:
        prefix_sum += num
        # ถ้ามี prefix_sum - k เคยเจอมาก่อน แปลว่ามี subarray ที่ผลรวม = k
        if (prefix_sum - k) in freq:
            count += freq[prefix_sum - k]
        freq[prefix_sum] = freq.get(prefix_sum, 0) + 1

    return count

print(subarray_sum([1, 1, 1], 2))       # 2
print(subarray_sum([1, -1, 0], 0))      # 3
```

### 11. LRU Cache Implementation
Time: O(1) ทั้ง get/put — doubly linked list ทำให้ลบ/แทรกได้ O(1), hash map ทำให้หา node ได้ O(1)

Space: O(capacity)

```python
class Node:
    def __init__(self, key=0, val=0):
        self.key, self.val = key, val
        self.prev = self.next = None

class LRUCache:
    def __init__(self, capacity):
        self.cap = capacity
        self.map = {}  # key -> Node
        self.head = Node()  # dummy most-recent side
        self.tail = Node()  # dummy least-recent side
        self.head.next = self.tail
        self.tail.prev = self.head

    def _remove(self, node):
        node.prev.next = node.next
        node.next.prev = node.prev

    def _insert_front(self, node):  # ใส่ไว้หลัง head (recently used)
        node.next = self.head.next
        node.prev = self.head
        self.head.next.prev = node
        self.head.next = node

    def get(self, key):
        if key not in self.map:
            return -1
        node = self.map[key]
        self._remove(node)
        self._insert_front(node)
        return node.val

    def put(self, key, value):
        if key in self.map:
            self._remove(self.map[key])
        node = Node(key, value)
        self.map[key] = node
        self._insert_front(node)

        if len(self.map) > self.cap:
            lru = self.tail.prev
            self._remove(lru)
            del self.map[lru.key]

cache = LRUCache(2)
cache.put(1, 1)
cache.put(2, 2)
print(cache.get(1))  # 1
cache.put(3, 3)       # evict key 2
print(cache.get(2))  # -1
```

### 12. Boyer-Moore Horspool String Matching
Time: O(N+M) เฉลี่ย (worst case O(NM) แต่พบไม่บ่อย)

Space: O(K) — K คือขนาด alphabet

```python
def horspool_search(text, pattern):
    n, m = len(text), len(pattern)
    if m == 0 or m > n:
        return -1

    # สร้าง Bad Character Shift Table
    shift = {ch: m for ch in set(text)}
    for i in range(m - 1):  # ไม่รวมตัวสุดท้าย
        shift[pattern[i]] = m - 1 - i

    i = 0
    while i <= n - m:
        j = m - 1
        while j >= 0 and text[i + j] == pattern[j]:
            j -= 1
        if j < 0:
            return i  # เจอ pattern
        bad_char = text[i + m - 1]
        i += shift.get(bad_char, m)

    return -1

print(horspool_search("JIM_SAW_ME_IN_A_BARBERSHOP", "BARBER"))  # 19
```

### 13. Palindrome Substrings (Lookup Table)
Time: O(N²) — สอง loop ซ้อนกันไล่ทุกช่วง (i, j)

Space: O(N²) — 2D table

```python
def count_palindrome_substrings(s):
    n = len(s)
    dp = [[False] * n for _ in range(n)]
    count = 0

    for i in range(n):        # ความยาว 1 เป็น palindrome เสมอ
        dp[i][i] = True
        count += 1

    for length in range(2, n + 1):        # ไล่ตามความยาว substring
        for i in range(n - length + 1):
            j = i + length - 1
            if s[i] == s[j]:
                if length == 2 or dp[i+1][j-1]:
                    dp[i][j] = True
                    count += 1

    return count

print(count_palindrome_substrings("aaa"))  # 6 -> a,a,a,aa,aa,aaa
```

### 14. Word Break Problem
Time: O(N²) — สอง loop ซ้อนกัน (i, j) ในกรณีที่การ slice+lookup ถือเป็น O(1) โดยประมาณ (จริงๆ slicing มีต้นทุน O(N) เพิ่มเติม แต่โดย pattern การนับ complexity ในข้อสอบมักตัดจบที่ O(N²))

Space: O(N) — dp array + O(sum of word lengths) สำหรับ word_set

```python
def word_break(s, word_dict):
    n = len(s)
    word_set = set(word_dict)
    dp = [False] * (n + 1)
    dp[0] = True  # substring ว่าง = แบ่งได้เสมอ

    for i in range(1, n + 1):
        for j in range(i):
            if dp[j] and s[j:i] in word_set:
                dp[i] = True
                break

    return dp[n]

print(word_break("leetcode", ["leet", "code"]))  # True
```

### 15. Trie Tree for Autocomplete
Time: O(L) ต่อ insert/search โดย L คือความยาวคำ

Space: O(จำนวนคำ × ความยาวเฉลี่ย × alphabet size) — ใช้เยอะแต่แลกความเร็ว

```python
class TrieNode:
    def __init__(self):
        self.children = {}
        self.is_end = False

class Trie:
    def __init__(self):
        self.root = TrieNode()

    def insert(self, word):
        node = self.root
        for ch in word:
            if ch not in node.children:
                node.children[ch] = TrieNode()
            node = node.children[ch]
        node.is_end = True

    def _find_node(self, prefix):
        node = self.root
        for ch in prefix:
            if ch not in node.children:
                return None
            node = node.children[ch]
        return node

    def search(self, word):
        node = self._find_node(word)
        return node is not None and node.is_end

    def starts_with(self, prefix):
        return self._find_node(prefix) is not None

trie = Trie()
for w in ["cat", "car", "card", "care"]:
    trie.insert(w)
print(trie.starts_with("car"))  # True
print(trie.search("ca"))        # False
```

### 16. Rabin-Karp Algorithm (Rolling Hash)
Time: O(N+M) เฉลี่ย (การ rolling hash ทำให้แต่ละ shift ใช้ O(1))

Space: O(1) — ไม่นับ input

```python
def rabin_karp(text, pattern, base=256, mod=10**9 + 7):
    n, m = len(text), len(pattern)
    if m > n:
        return -1

    h = pow(base, m - 1, mod)  # base^(m-1) mod mod
    p_hash = t_hash = 0

    for i in range(m):
        p_hash = (p_hash * base + ord(pattern[i])) % mod
        t_hash = (t_hash * base + ord(text[i])) % mod

    for i in range(n - m + 1):
        if p_hash == t_hash:
            if text[i:i+m] == pattern:  # กัน hash collision
                return i
        if i < n - m:
            # rolling hash: ตัดตัวหน้าออก เลื่อนเข้าตัวใหม่
            t_hash = ((t_hash - ord(text[i]) * h) * base + ord(text[i + m])) % mod

    return -1

print(rabin_karp("ABABDABACDABABCABAB", "ABABCABAB"))  # 10
```

### 17. Min Stack $O(1)$
Time: O(1) ทุกฟังก์ชัน

Space: O(N) — auxiliary stack ขนาดเท่ากับ main stack

```python
class MinStack:
    def __init__(self):
        self.stack = []
        self.min_stack = []  # เก็บค่า min ณ แต่ละชั้น

    def push(self, val):
        self.stack.append(val)
        current_min = min(val, self.min_stack[-1]) if self.min_stack else val
        self.min_stack.append(current_min)

    def pop(self):
        self.stack.pop()
        self.min_stack.pop()

    def top(self):
        return self.stack[-1]

    def get_min(self):
        return self.min_stack[-1]

ms = MinStack()
ms.push(3); ms.push(1); ms.push(2)
print(ms.get_min())  # 1
ms.pop()
print(ms.get_min())  # 1 (ยังเหลือ 3,1)
```

### 18. Find All Duplicates in an Array
Time: O(N) — วนลูปครั้งเดียว

Space: O(1) นอกเหนือจาก output list — ใช้ array เดิมเป็น hash table (ค่าตั้งแต่ 1 ถึง N ใช้เป็น index ได้พอดี)

```python
def find_all_duplicates(nums):
    result = []
    for num in nums:
        idx = abs(num) - 1
        if nums[idx] < 0:
            result.append(abs(num))  # เคย mark แล้ว = ซ้ำ
        else:
            nums[idx] = -nums[idx]  # mark ว่าเจอแล้ว

    return result

print(find_all_duplicates([4, 3, 2, 7, 8, 2, 3, 1]))  # [2, 3]
```

### 19. Climbing Stairs with Tabulation
Time: O(N) — loop เดียว bottom-up

Space: O(N) — dp array (ปรับให้เหลือ O(1) ได้โดยเก็บแค่ 2 ตัวแปรล่าสุด)

```python
def climb_stairs(n):
    if n <= 2:
        return n

    dp = [0] * (n + 1)
    dp[1], dp[2] = 1, 2

    for i in range(3, n + 1):
        dp[i] = dp[i - 1] + dp[i - 2]

    return dp[n]

print(climb_stairs(5))  # 8
```

### 20. Sparse Matrix Representation
Time: O(K) สำหรับการบวก โดย K คือจำนวนสมาชิกที่ไม่ใช่ 0 (แทนที่จะเป็น O(MN))

Space: O(K) แทนที่จะเป็น O(MN) — ประหยัดมหาศาลเมื่อ matrix เป็น sparse

```python
class SparseMatrix:
    def __init__(self, rows, cols):
        self.rows, self.cols = rows, cols
        self.data = {}  # (r, c) -> value  (เก็บเฉพาะที่ไม่ใช่ 0)

    def set(self, r, c, val):
        if val != 0:
            self.data[(r, c)] = val
        elif (r, c) in self.data:
            del self.data[(r, c)]

    def get(self, r, c):
        return self.data.get((r, c), 0)

    def add(self, other):
        result = SparseMatrix(self.rows, self.cols)
        result.data = dict(self.data)  # copy ค่าตั้งต้น
        for pos, val in other.data.items():
            result.data[pos] = result.data.get(pos, 0) + val
            if result.data[pos] == 0:
                del result.data[pos]
        return result

m1 = SparseMatrix(10000, 10000)
m1.set(0, 0, 5)
m1.set(9999, 9999, 3)
m2 = SparseMatrix(10000, 10000)
m2.set(0, 0, 2)

result = m1.add(m2)
print(result.get(0, 0))       # 7
print(len(result.data))       # 2 (ไม่ใช่ 100,000,000!)
```
