# Exception

ArithmeticError = ข้อผิดพลาดเกี่ยวกับการคำนวณ             
         
FileNotFoundError = เกิดเมื่อ เปิดไฟล์แล้วแต่ไฟลล์ไม่มี                                                          

NameError = ข้อผิดพลาดเกี่ยวกับชื่อกับตัวแปร                           
         
IndexError = เกิดเมื่อถามหาตำแหน่งแล้วไม่มี (list)         

TypeError = ข้อผิดพลาดเกี่ยวกับประเภทข้อมูล                                  

KeyError = ใช้กับ dictionary เกิดขึ้นเมื่อไม่มี key 

ValueError = ข้อผิดพลาดเกี่ยวกับข้อมูลไม่ตรงตามประเภทที่ระบุ

ZeroDivisionError = ข้อผิดพลาดเกี่ยวกับการหารด้วยศูนย์

try = ลองทำคำสั่งปกติ

expect = ถ้าเกิดข้อผิดพลาดจะมาทำงานตรงนี้

finally = จะเกิดข้อผิดพลาดหรือไม่เกิด ก็จะทำงานส่วนตรงนี้ทั้งครั้ง ซึ่งคำสั่งที่ ระบุมักจะเป็นคำสั่งสำคัญของโปรแกรม

```python
try:
    n1 = int(input("ตัวเลขตัวที่ 1 : "))
    n2 = int(input("ตัวเลขตัวที่ 2 : "))
    result = n1/n2
    print(result)

except ValueError:
    print("ข้อมูลไม่ถูกต้อง กรุณาป้อนข้อมูลเฉพาะตัวเลข")

except ZeroDivisionError:
    print("หารด้วย 0 ไม่ได้ เนื่องจากไม่ถูกนิยามทางคณิตศาสตร์")

finally:
    print("-"*12)
    print("จบโปรแกรม")
    print("-"*12)
```

```python
try:
    n1 = int(input("ตัวเลขตัวที่ 1 : "))
    n2 = int(input("ตัวเลขตัวที่ 2 : "))
    result = n1/n2
    if n1 < 0 or n2 < 0:
        raise Exception("ป้อนตัวเลขที่เป็นค่าบวกเท่านั้น!!!") #raise สร้างข้อผิดพลาดขึ้นมาเอง

except ZeroDivisionError as e:
    print(f"เกิดข้อผิดพลาด : {e}")

else: #ใช้ else เมื่อไม่เกิดข้อผิดพลาด
    print(f"ผลลัพธ์สำเร็จ : {result}")

finally:
    print("-"*12)
    print("จบโปรแกรม")
    print("-"*12)
```
