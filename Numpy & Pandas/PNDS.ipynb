# Pandas

## Pandas เป็นไลบราลี่ภาษา python สำหรับจัดการและวิเคราะห์ข้อมูลที่เป็นแบบโครงสร้างทั้งรูปแบบมิติเดียวกันและหลายมิติ
## โครงสร้างของ Pandas จะมี Series, DataFrame, Panel
### Series คือ เป็นโครงสร้างข้อมูลแบบ Array 1มิติ ที่เก็ฐข้อมูลต่างชนิดได้ มีส่วนประกอบ 2 ส่วน คือ
- Index ใช้อ้างอิงตำแหน่งของข้อมูล
- Element Value คือ ข้อมูลในแต่ละ Index

### DataFrame เป็นโครงสร้างข้อมูลแบบ Array 2มิติ (เป็นตารางประกอบด้วย 2 คอลัมขึ้นไป) หรือการนำ Series หลายๆอันมาต่อเรียงกัน เช่น ข้อมูลคะแนนนักเรียน ข้อมูลสินค้า หรือข้อมูลประชากร เป็นต้น

### Panel เป็นโครงสร้างจ้อมูลแบบ 3 มิติ เป็นการนำ DataFrame หลายๆชั้น ประกอบไปด้วย 3 แกน
- แกนที่ 0 - DataFrame
- แกนที่ 1 - แถวหรือ Index ของแต่ละ DataFrame
- แกนที่ 2 - คอลัมของแต่ละ DataFrame

## หัวข้อที่ 1 การสร้าง Series จาก List และ Tuple

```python
import pandas as pd

data_ls = [10, 20, "Film", 15.05, "มะละกิ"] ##List
data_tp = (10, 20, "Film", 15.05, "มะละกอ") ##Tuple
sr = pd.Series(data_ls)
sr1 = pd.Series(data_tp)
print(sr)
print(f"\n{sr1}")
```

## สร้าง Series จาก Numpy

## สร้าง Series แบบกำหนดเลข index

## การเข้าถึงขอมูลใน Series (แบบช่วงข้อมูล Slice)
### โครงสร้างการเข้าถึง คือ
- ps[start:stop-1]

### สัญลักษณ์
- [:] คือ ทุกอัน
- [2:] คือ index ที่ 2 เป็นต้นไป
- [:2] คือ เริ่มตันไปจนก่อนถึง 2
- [1:3] คือ เริ่มต้นที่ index 1 ไปก่อนถึง index 3

```python
import pandas as pd

ps = pd.Series([10, 20, 30, 40, 50, 60, 70, 80])
ps
##ต้องการตัวเลขที่มีค่ามากกว่าหรือเท่ากับ 40 จะทำได้โดย
print(f"มีค่ามากกว่า 40 \n{ps[3:]}")
print(f"\nเอาตัวเลขทุกตัวก่อนถึง index 4 = \n{ps[:4]}") #งั้นก็ว่า แสดงวิธีการหาโดย โดยนับจากเลข 0 ไปถึงยีงเห้องเลขที่มีค่าน้อยกว่า 1 index
```

## การสร้าง DataFrame จาก List ด้วย zip

```python
import pandas as pd

#zip คือ การบีบอัดข้อมูล
#ต้องทำเป็นแบบเรียงชื่อกัน ภายใน list เลย เช่น list นี้เป็นชื่อ ก็ต้องเป็นชื่อให้หมด แล้วค่อยไปหัวข้อต่อไป
data1 = ["Cookie run classic", "Arena of valor", "PUBG"]
data2 = ["3M Downloads", "20M Downloads", "5M Downloads"]
data3 = ["First Chart", "Second Chart", "Tenth Chart"]

dtzip = list(zip(data1,data2,data3))
cols = ["Name", "Number of Downloads", "Number of Chart"]

ps = pd.DataFrame(dtzip, columns=cols)
ps
```

## สร้าง DataFrame จาก Series

## ตรวจสอบข้อมูล ใน DataFrame
- head() == อ่านข้อมูล 5 แถวแรก

- head(n) == อ่านข้อมูล n แถวแรก

- tail() == อ่านข้อมูล 5 แถวสุดท้าย

- tail(n) == อ่านข้อมูล n แถวสุดท้าย

- sample(n) == สุ่มจำนวนข้อมูล n แถว

- describe() == ดูค่าทางสถิติ เช่น ค่าต่ำสุด-สูงสุด ค่าเฉลี่ยข้อมูล เป็นต้น
    
    ดูค่าทางสถิติเบื้องต้น
    - mean() = ค่าเฉลี่ย
    - max() = ค่าสูงสุด
    - min() = ค่าน้อยสุด
    - count() = นับจำนวนข้อมูลทั้งหมด
    - std() = ส่วนเบี่ยงเบนมาตรฐาน
    - sum() = ผลรวมข้อมูล

- shape == สำหรับนับจำนวนข้อมูล เช่น (20, 4) คือ 20 แถว 4 คอลัมน์

- columns == ตรวจสอบว่ามีคอลัมน์อะไรบ้าง

- values == อ่านข้อมูลแบบ Array

## ค้นหาข้อมูลด้วย Math & Contains
- match(string) ค้นหาคำโดย เริ่มจากอักขระตัวแรก

- contains(string) ค้นหาคำ ไม่จำเป็นต้องเริ่มจากอักขระตัวแรก

```python
import pandas as pd

df = pd.read_excel("dataupdate.xlsx", sheet_name="weather", index_col="Day")# กำหนดให้ Day เป็น index ของ DataFrame
df.Event.str.match("ฝนตก") #คือ การตรวจสอบว่าในคอลัมน์ Event มีคำว่า "ฝนตก" หรือไม่ โดยจะคืนค่าเป็น Series ของ Boolean (True/False) สำหรับแต่ละแถว ต้องมีให้ครบทุกตัวอักขระ
df[df.Event.str.contains("ตก")] #แค่มีบางคำก็ได้ แนะนำ เพราะส่งค่าออกมาทำให้สวยด้วย
```

## การลูปข้อมูล

มี Function ที่มีชื่อว่า iterrows() เพื่อทำการลูป ก็จะได้ key กับ value มา

```python
import pandas as pd

df = pd.read_excel("dataupdate.xlsx", sheet_name="weather", index_col="Day")
for idx, row in df.iterrows():
    print(f"อุณหภูมิ = {row.Temperature} ส่งผลให้เกิด {row.Event}")
```

## สอบถามข้อมูลแบบมีเงือนไข

```python
import pandas as pd

df = pd.read_excel("dataupdate.xlsx", sheet_name="weather", index_col="Day")
df[df.Temperature <= 20] #อยากทราบว่าวันใดบ้างที่มีอุณหภมิ น้อยกว่าหรือเท่ากับ 20
df[df.Event == "แดดร้อน"] #อยากทราบว่าวันใดบ้างที่มี อากาศ แดดร้อน
df.loc[(df.Event == "ฝนตก") & (df.Temperature >= 35)] #อยากทราบว่าวันใดบ้างที่มีสภาพอากาศ ฝนตก และ มีอุณหภูมิ มากกว่าหรีอเท่ากับ 35
df.loc[(df.Event == "เมฆมาก") | (df.Temperature <= 18)]
## df.loc[] คือ การรวมฟังก์ชัน
```

## การเลือกช่วงข้อมูลด้วย isin(List) ใช้แทน OR

```python
import pandas as pd

df = pd.read_excel("dataupdate.xlsx", sheet_name="weather", index_col="Day")
df.loc[(df.Temperature.isin([18, 20, 25, 35]))] #ต้องการเอาอุณหภูมิ 18, 20, 25, 35 หรือต้องการเอาข้อมูลหลายตัว จะใช้คำสั่งนี้ ครอบ list[] ให้ด้วย
df.loc[(df.Temperature.isin([20, 35]))] #ต้องการเอาอุณหภูมิ ที่มีค่า 20 และ 35 องศาเซลเซียส เท่านั้น
df.loc[(df.Temperature.isin([30, 35]))][df.Event == "แดดร้อน"] #ต้องการเอาอุณหภูมิ ที่มีค่า 30 และ 35 องศาเซลเซียส และ ต้องการเอาสภาพอากาศที่มี แดดร้อน
```
