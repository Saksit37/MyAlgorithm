7. ถ้าต้องออกแบบระบบเก็บใบรับรองการศึกษา ควรเก็บข้อมูลจริงทั้งหมดใน block หรือเก็บเฉพาะ hash ของข้อมูล เพราะเหตุใด

```python
import hashlib
import json
import time

class Block:
    def __init__(self, index, timestamp, data, previous_hash):
        self.index = index
        self.timestamp = timestamp
        self.data = data
        self.previous_hash = previous_hash
        self.hash = self.calculate_hash()

    def calculate_hash(self):
        block_content = {
            "index": self.index,
            "timestamp": self.timestamp,
            "data": self.data,
            "previous_hash": self.previous_hash
        }
        block_string = json.dumps(block_content, sort_keys=True)
        return hashlib.sha256(block_string.encode()).hexdigest()
    

class Blockchain:
    def __init__(self):
        self.chain = [self.create_genesis_block()]

    def create_genesis_block(self):
        return Block(
            index = 0,
            timestamp = time.time(),
            data = "Genesis Block",
            previous_hash = "0" 
        )

    def get_latest_block(self):
        return self.chain[-1]

    def add_block(self, data):
        latest_block = self.get_latest_block()
        new_block = Block(
            index = len(self.chain),
            timestamp = time.time(),
            data = data,
            previous_hash = latest_block.hash
        )
        self.chain.append(new_block)

    def is_chain_valid(self):
        for i in range(1, len(self.chain)):
            cr_block = self.chain[i]
            previousblock = self.chain[i-1]

            if cr_block.hash != cr_block.calculate_hash():
                return False
            
            if cr_block.previous_hash != previousblock.hash:
                return False
        
        return True

    def print_chain(self):
        for block in self.chain:
            print("---------------------------")
            print("Index :", block.index)
            print("Timestamp :", block.timestamp)
            print("Data :", block.data)
            print("Previous Hash :", block.previous_hash)
            print("Hash :", block.hash)
    


my_chain = Blockchain()



my_chain.add_block("มาวิน ยืมเงิน วิลเลี่ยม 700 บาท")
my_chain.add_block("วิลเลี่ยม ยืมเงิน พิมรี่พาย 2300 บาท")
my_chain.add_block("พิมรี่พาย ยืมเงิน แสงลม 100 บาท")
#Add เพิ่ม block 3 block

my_chain.add_block("ไปขอยืมหนังสือ")
my_chain.add_block("ต่อระยะเวลาการยืมหนังสือ")
my_chain.add_block("คืนหนังสือ")

# my_chain.chain[1].data = "มาวิน ยืมเงิน วิลเลี่ยม 999999 บาท"
# my_chain.chain[1].hash = my_chain.chain[1].calculate_hash()

my_chain.print_chain()

print("Is blockchain valid :", my_chain.is_chain_valid())
```

# รูปภาพหรือเพิ่มจุดหลัง TX2

1. hash ของ TX2 เปลี่ยนหรือไม่
- เปลี่ยน เพราะข้อมูลของ TX2 ถูกแก้ไข ถึงแม้ว่าจะแค่นิดเดียว hash ก็จะเปลี่ยนทันที

2. parent hash ที่เกี่ยวข้องกับ TX2 เปลี่ยนหรือไม่
- เปลี่ยน เพราะว่า parent hash นั้นได้มาจากการคำนวณของ TX1 และ TX2 ซึ่งเมื่อเกิดการเปลี่ยนแปลงของ TX ใด TX หนึ่ง ก็จะเกิดการคำนวณใหม่ จึงทำให้ parent hash เปลี่ยนตาม

3. Merkle root เปลี่ยนหรือไม่
- เปลี่ยน เพราะ Merkle Root ถูกสร้างจาก parent hash ทั้งสองฝั่งเมื่อ parent hash ฝั่งใดแก้ไขหรือเกิดการเปลี่ยน Merkle Root ก็เปลี่ยนตาม

4. hash ของฝั่ง TX3 และ TX4 เปลี่ยนหรือไม่
- ไม่เปลี่ยน เพราะไม่ได้แก้ไขข้อมูลของ TX3 และ TX4 ทำให้ Hash ของฝั่งนี้ยังเป็นค่าเดิม

5. การทดลองนี้บอกอะไรเกี่ยวกับความไวของ hash และ Merkle root ต่อการเปลี่ยนข้อมูล
- จากที่ผมได้ลองทำการแก้ไขข้อมูลของ TX2 ก็เห็นได้ว่า hash เปลี่ยนทันที และยังส่งผลกระทบไปยัง parent hash และ Merkle root ต้องเปลี่ยนไปอีกด้วย ทำให้ได้รู้ว่า hash มีความไวต่อการเปลี่ยนแปลงข้อมูลที่มาก แม้จะแก้ไขแค่นิดเดียว

# Part C : Explore Proof Path
ให้นิสิตใช้ Blockchain Academy Merkle Tree Simulator เพื่อดู proof path ของ data item ที่เลือก เช่น TX2

ให้นิสิตเลือก TX2 แล้วสังเกต node ที่เกี่ยวข้องกับการพิสูจน์ว่า TX2 อยู่ใน Merkle tree จากนั้นตอบคำถามต่อไปนี้
1. ถ้าต้องพิสูจน์ว่า TX2 อยู่ใน Merkle tree ต้องใช้ sibling hash ตัวใดในระดับแรก
- ต้องใช้ตัว hash ของ TX1 เป็น sibling hash

2. หลังจากรวม TX2 กับ sibling hash แล้ว ต้องใช้ hash ตัวใดต่อเพื่อไปถึง Merkle root
- หลังจากที่รวม TX1 และ TX2 เป็น parent hash แล้ว ก็ต้องนำไป hash ร่วมกับ parent hash ของ TX3 และ TX4 เพื่อคำนวณเป็น Merkle root

3. ใน proof path จำเป็นต้องใช้ข้อมูลของ transaction ทุกตัวหรือไม่
- ไม่ ใช้แค่ Hash ที่อยู่ในเส้นของ Transaction นั้นก็จะสามารถตรวจสอบได้แล้ว

4. ทำไมการรู้เฉพาะ sibling hash บางตัวจึงเพียงพอสำหรับการ reconstruct Merkle root
- เพราะ Sibling hash ที่อยู่ในแต่ละระดับก็เพียงพอที่จะคำนวณต่อไปจนถึง Merkle Root ได้ เลยไม่ต้องใช้ข้อมูลทั้งหมด

5. ถ้าสลับลำดับซ้าย/ขวาของ hash ผลการคำนวณจะยังตรงกับ Merkle root เดิมหรือไม่
- ไม่ตรง จากที่ศึกษามา การคำนวณ hash จะใช้ลำดับซ้ายและขวาตามเดิม ถ้าสลับตำแหน่งกัน ค่า hash ที่ได้ก็จะเปลี่ยน เลยทำให้ Merkle root ไม่ตรงกัน 
เช่น hash(a) + hash(b) ไม่เท่ากับ hash(b) + hash(a) **ลำดับเปลี่ยน ผลลัพธ์เปลี่ยน**
