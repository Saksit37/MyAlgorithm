### กราฟแท่ง

```python
import matplotlib.pyplot as plt

x = [1, 2, 3, 4, 5] ##แนวนอน
y = [10, 20, 30, 40, 50] ##แนวตั้ง

plt.bar(x,y) ##กราฟแท่ง
plt.show() ##แสดงกราฟ
```

### กราฟเส้น

```python
import matplotlib.pyplot as plt

product1 = [10, 20, 30, 40, 50] ##ยอดขายสินค้าที่ 1
product2 = [15, 30, 10, 20, 60] ##ยอดขายสินค้าที่ 2 
month = [1, 2, 3, 4, 5] ##เดือน

plt.plot(month,product1) ##กราฟเส้น 1 #เอาเดือนมาแสดงผลเป็นแนวนอน
plt.plot(month,product2) ##กราฟเส้น 2
plt.show() ##แสดงกราฟ

#น้ำเงิน = product1
#ส้ม = product2
```

## Xlabel และ Ylabel เพื่ออธิบายข้อมูลในแนวตั้งและนอน

```python
import matplotlib.pyplot as plt

product1 = [10, 20, 30, 40, 50] ##ยอดขายสินค้าที่ 1
product2 = [15, 30, 10, 20, 60] ##ยอดขายสินค้าที่ 2 
month = [1, 2, 3, 4, 5] ##เดือน

plt.plot(month,product1) ##กราฟเส้น 1 #เอาเดือนมาแสดงผลเป็นแนวนอน
plt.plot(month,product2) ##กราฟเส้น 2
plt.xlabel("Month") ##แสดงชื่อว่าแนวนอนนั้นมีชื่อว่าอะไร
plt.ylabel("Product")
plt.show() ##แสดงกราฟ

#น้ำเงิน = product1
#ส้ม = product2
```

## การกำหนดสี

```python
import matplotlib.pyplot as plt

product1 = [10, 20, 30, 40, 50] ##ยอดขายสินค้าที่ 1
product2 = [15, 30, 10, 20, 60] ##ยอดขายสินค้าที่ 2 
month = [1, 2, 3, 4, 5] ##เดือน

plt.plot(month, product1, color="g") ##กราฟเส้น 1 #เอาเดือนมาแสดงผลเป็นแนวนอน
plt.plot(month, product2, color="m") ##กราฟเส้น 2
plt.xlabel("Month") ##แสดงชื่อว่าแนวนอนนั้นมีชื่อว่าอะไร
plt.ylabel("Product")
plt.show() ##แสดงกราฟ

###
# b = blue, g = green, r = red, c = cyan, m = magenta, y = yellow, k = black, w = white
# ###
```

## กำหนดจุด Maker บน กราฟ
Marker Support
- "." = point
 
- "o" = cycle

- "x" = x marker

- "+" = plus

- "*" = star

- "s" = square

- "|" = vline

- "_" = hline

- "v" = down

- "^" = up

- ">" = right

- "<" = left

```python
import matplotlib.pyplot as plt

product1 = [10, 20, 30, 40, 50] ##ยอดขายสินค้าที่ 1
product2 = [15, 30, 10, 20, 60] ##ยอดขายสินค้าที่ 2 
month = [1, 2, 3, 4, 5] ##เดือน

plt.plot(month, product1, color="g", marker="s") ##กราฟเส้น 1 #เอาเดือนมาแสดงผลเป็นแนวนอน
plt.plot(month, product2, color="m", marker="+") ##กราฟเส้น 2
plt.xlabel("Month") ##แสดงชื่อว่าแนวนอนนั้นมีชื่อว่าอะไร
plt.ylabel("Product")
plt.show() ##แสดงกราฟ
```

### กำหนดรูปแบบเส้น Line Support
- "-" = solid

- "--" = dashed

- "-." = dash-dot

- ":" = dotted

```python
import matplotlib.pyplot as plt

product1 = [10, 20, 30, 40, 50] ##ยอดขายสินค้าที่ 1
month = [1, 2, 3, 4, 5] ##เดือน

plt.plot(month, product1, color="g", marker="s", linestyle="-.")
plt.xlabel("Month") ##แสดงชื่อว่าแนวนอนนั้นมีชื่อว่าอะไร
plt.ylabel("Product")
plt.show() ##แสดงกราฟ
```

### การกำหนดหัวข้อด้วย title

```python
import matplotlib.pyplot as plt

product1 = [10, 20, 30, 40, 50] ##ยอดขายสินค้าที่ 1
month = [1, 2, 3, 4, 5] ##เดือน

## หรือจะสร้างไว้ก่อนกะได้ ผ่าน dictionary
data = {"size" : 15, "color" : "green", "backgroundcolor" : "cyan"}

plt.plot(month, product1, color="g", marker="s", linestyle="-.")
plt.xlabel("Month", data) ##แสดงชื่อว่าแนวนอนนั้นมีชื่อว่าอะไร
plt.ylabel("Product", data)

config = {"size" : 15, "color" : "green"} #กำหนดขนาด สี
plt.title("Top Product 2026", config, loc="right") #กำหนดหัวข้อ loc คือ ตำแหน่ง อยากไปไหนก็แล้วแต่ ==ขวา==
# plt.title("Top Product 2026", config, loc="left") #กำหนดหัวข้อ loc คือ ตำแหน่ง อยากไปไหนก็แล้วแต่ ==ซ้าย==
# plt.title("Top Product 2026", config, loc="center") #กำหนดหัวข้อ loc คือ ตำแหน่ง อยากไปไหนก็แล้วแต่ ==กลาง==

plt.show() ##แสดงกราฟ
```

### Legend Function

```python
import matplotlib.pyplot as plt

product1 = [10, 20, 30, 40, 50] ##ยอดขายสินค้าที่ 1
product2 = [5, 10, 15, 20, 25]
month = [1, 2, 3, 4, 5] ##เดือน

## หรือจะสร้างไว้ก่อนกะได้ ผ่าน dictionary
data = {"size" : 15, "color" : "green", "backgroundcolor" : "cyan"}

plt.plot(month, product1, color="g")
plt.plot(month, product2, color="r")
plt.xlabel("Month", data) ##แสดงชื่อว่าแนวนอนนั้นมีชื่อว่าอะไร
plt.ylabel("Product", data)

config = {"size" : 15, "color" : "green"} #กำหนดขนาด สี
plt.title("Top Product 2026", config, loc="right") #กำหนดหัวข้อ loc คือ ตำแหน่ง อยากไปไหนก็แล้วแต่ ==ขวา==
# plt.title("Top Product 2026", config, loc="left") #กำหนดหัวข้อ loc คือ ตำแหน่ง อยากไปไหนก็แล้วแต่ ==ซ้าย==
# plt.title("Top Product 2026", config, loc="center") #กำหนดหัวข้อ loc คือ ตำแหน่ง อยากไปไหนก็แล้วแต่ ==กลาง==

plt.legend(["product1","product2"])
```

### กำหนดข้อความ Legend
